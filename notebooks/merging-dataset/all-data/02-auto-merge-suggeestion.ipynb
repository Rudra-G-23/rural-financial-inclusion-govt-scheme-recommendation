```python
def load_dataset(path, df_name):
    df_name = pd.read_csv(path)
    return df_name
```

```python
ROOT_DIR = r"C:\Users\Rudra\Desktop\rural-financial-inclusion-govt-scheme-recommendation\sample-csv"
```

```python
datasets = []

for i in range (1, 16):
    
    level = f"{i:02d}"
    name = f"df_{i:02d}"
    
    path = os.path.join(ROOT_DIR, f"sample-lev-{level}.csv")
    name = load_dataset(path=path, df_name=name)
    
    datasets.append(name)
```

```python
datasets
```

```python
print("--- Comparison Matrix ---")
matrix = fmj.compare(datasets)
print(matrix.head())
```

```python
matrix.to_csv("matrix.csv")
```
