```python
def load_plot(path, cols:list, lev, title:str = None, ) -> None:
    
    path = path    
    pdf = pl.read_parquet(path)
    pdf = pdf[cols]
    corr = pdf.corr()
    mask = np.triu(np.ones_like(corr, dtype=bool))

    display(pdf.head(3))
    display(pdf.null_count())
    display(corr)
    
    plt.figure(figsize=(10, 8))
    sns.heatmap(
        corr,
        annot=True,
        cmap='viridis',
        center=0,
        cbar=True,
        mask=mask,
        xticklabels=pdf.columns,
        yticklabels=pdf.columns,
    )
    plt.title(f"{title} Correlation Matrix Heatmap)")
    plt.show()

    save_path = fr"C:\Users\Rudra\Desktop\rural-financial-inclusion-govt-scheme-recommendation\notebooks\correlations-all-levels\corr\cor-{lev}.parquet"
    pdf.write_parquet(save_path)
    print(f"Parquet saved to: {save_path}")
    
    del pdf
```
