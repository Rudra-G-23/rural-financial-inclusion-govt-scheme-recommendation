# Useful Variables

```python
lev_02_hh = [ 'Revisit_Status', 'FDQ_Original_Member', 'Multiplier']

lev_02_food = ['Meals_Usually_Taken_Per_Day','Meals_From_School',
               'Meals_From_Employer','Meals_Other',
               'Meals_On_Payment','Meals_At_Home']

lev_02_digital = ['Used_Internet_Last_30_Days']

lev_02__behavior = ['Days_Away_From_Home_Last_30_Days']

lev_02_user = ['FSU_Serial_No', 'Relation_to_Head','Gender',
               'Age', 'Marital_Status', 
               'Education_Level', 'Years_of_Education']
```

```python
lev_02_cols = (
    lev_02_hh
    + lev_02_food
    + lev_02_digital
    + lev_02__behavior
    + lev_02_user
)
```

```python
df = pdf.select(lev_02_cols)
```

```python
df
```

```python
df.head(2).collect()
```

```python
df = df.with_columns(
    [pl.col(col).cast(pl.Int32, strict=False) for col in lev_02_cols]
)
```

```python
unique_counts = df.select(pl.all().n_unique()).collect()
unique_counts
```

```python
unique_counts = df.select(pl.all().n_unique()).collect()

categorical_cols = []
numerical_cols = []

for col in df.columns:
    if unique_counts[col][0] <13:
        categorical_cols.append(col)
    else:
        numerical_cols.append(col)
```

```python
stats = df.select(numerical_cols).describe()
with pl.Config(tbl_rows=-1, tbl_cols=-1):
    display(stats.to_pandas().T)
```

```python
for col in categorical_cols:
    print(col)
    counts = df.select(pl.col(col).value_counts(sort=True)).collect()
    
    with pl.Config(tbl_rows=-1, tbl_cols=-1):
        display(counts.unnest(col))
```
