# Useful Variables

```python
lev_04_fin = ['Ration_Any_Item_Last_30_Days' ,'Ration_Rice',
              'Ration_Wheat','Ration_Coarse_Grain',
              'Ration_Sugar','Ration_Pulses',
              'Ration_Edible_Oil','Ration_Other_Food_Items']

lev_04_digital = [
    'Online_Groceries',
'Online_Milk',
'Online_Vegetables',
'Online_Fresh_Fruits',
'Online_Dry_Fruits',
'Online_Egg_Fish_Meat',
'Online_Served_Processed_Food',
'Online_Packed_Processed_Food',
'Online_Other_Food_Items',
]

lev_04_behavior = ['Ceremony_Performed_Last_30_Days']
lev_04_food = ['Meals_Served_to_Non_HH_Members']

lev_04_cols = (
    lev_04_food 
    + lev_04_behavior
    + lev_04_digital 
    + lev_04_fin
)
```

```python
df = pdf.select(lev_04_cols)
```

```python
df.head(2).collect()
```

```python
df = df.with_columns(
    [pl.col(col).cast(pl.Int32, strict=False) for col in lev_04_cols]
)
```

```python
unique_counts = df.select(pl.all().n_unique()).collect()
unique_counts
```

```python
unique_counts = df.select(pl.all().n_unique()).collect()

categorical_cols = []
numerical_cols = []

for col in df.columns:
    if unique_counts[col][0] <13:
        categorical_cols.append(col)
    else:
        numerical_cols.append(col)
```

```python
stats = df.select(numerical_cols).describe()
with pl.Config(tbl_rows=-1, tbl_cols=-1):
    display(stats.to_pandas().T)
```

```python
for col in categorical_cols:
    print(col)
    counts = df.select(pl.col(col).value_counts(sort=True)).collect()
    
    with pl.Config(tbl_rows=-1, tbl_cols=-1):
        display(counts.unnest(col))
```
